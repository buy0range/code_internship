# extract images from video

```python
import cv2
import os

def video_to_frames(video_path, output_dir, interval, freq, force):
    """
    将视频按固定间隔抽帧（默认每30帧抽取1帧，假设视频为30fps，即每秒1帧）
    :param video_path: 视频文件路径（如 "input.mp4"）
    :param output_dir: 输出图片的目录（如 "output_frames"）
    :param interval: 抽帧间隔（每隔多少帧抽1帧）
    """
    # 创建输出目录
    os.makedirs(output_dir, exist_ok=True)
    
    # 打开视频文件
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print("错误：无法打开视频文件")
        return
    
    # 读取视频帧并保存
    frame_count = 0
    saved_count = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break  # 视频结束
        
        # 按间隔抽帧
        if frame_count % interval == 0:
            output_path = os.path.join(output_dir, f"{freq}Hz_{force}N_{saved_count:04d}.jpg")
            cv2.imwrite(output_path, frame)
            saved_count += 1
        
        frame_count += 1
    
    # 释放资源
    cap.release()
    print(f"视频 {video_path} 分解完成，共保存 {saved_count} 帧到 {output_dir}")
```

```python
video_path = r"D:\STUDY\internship\raw_data\100µm_10Hz_15N\100µm_10H_15N.avi"  # 替换为你的视频路径
output_dir = r"D:\STUDY\internship\raw_data\images"   # 输出目录
video_to_frames(video_path, output_dir, 1, 10, 15)  
```
