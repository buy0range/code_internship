# extract images from video

```python
import cv2
import os

def video_to_frames(video_path, output_dir, freq, force):
    """
    :param video_path: the path of the video
    :param output_dir: the path of the output directory
    :param freq: the frequency of the experiment
    :param force: the normal force of the experiment
    """
    # create output directory
    os.makedirs(output_dir, exist_ok=True)
    
    # open video
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print("can not open video file")
        return
    
    # read video and save frames
    frame_count = 0
    saved_count = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break  # video is over
        
        
        output_path = os.path.join(output_dir, f"{freq}Hz_{force}N_{saved_count:04d}.jpg")
        cv2.imwrite(output_path, frame)
        saved_count += 1
        
        frame_count += 1
    
    # release resources
    cap.release()
    print(f"video {video_path} has been split, {saved_count} has been saved to {output_dir}")
```

```python
video_path = r"D:\STUDY\internship\raw_data\100µm_10Hz_5N\100_µm_10_Hz_5_N.avi"
output_dir = r"D:\STUDY\internship\raw_data\images"   
video_to_frames(video_path, output_dir, 10, 5)
```
