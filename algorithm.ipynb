```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data_all.csv')
df.head()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 划分数据集
X = df.drop(['stage', 'material', 'stroke'], axis = 1)
y = df['stage']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y)
```

```python
print(X_train.shape)
```

```python
# select model
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, BatchNormalization, Dropout, Input  # 新增Input层

model = Sequential([
    Input(shape=(X_train.shape[1],)),  # 显式声明输入层
    Dense(256, activation='relu'),     # 移除input_shape参数
    BatchNormalization(),
    Dropout(0.3),
    Dense(128, activation='relu'),
    Dropout(0.2),
    Dense(64, activation='relu'),
    Dense(len(np.unique(y)), activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
print("Unique labels in y_train:", np.unique(y_train))
print("Number of classes (len(np.unique(y)):", len(np.unique(y)))
print("Max label in y_train:", y_train.max())
print("Min label in y_train:", y_train.min())
```

```python
# train model
model.fit(X_train, y_train, epochs=100, batch_size=32, validation_data=(X_test, y_test))

# evaluate model
score = model.evaluate(X_test, y_test, verbose=0)
print('Test loss:', score[0])
print('Test accuracy:', score[1])
```
