```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('original_data.csv', index_col=0)
df.head()
```

```python
df2 = df[['Time', 'coe', 'Absolute_Energy']].copy()
```

```python
df3 = df2.query("Time <= 198 and coe >= 0").copy()
df3.head()
```

```python
df3['culmative_Eabs'] = df3['Absolute_Energy'].cumsum()
```

```python
df3.tail()
```

```python
plt.plot(df3['Time'], df3['culmative_Eabs'])
```
